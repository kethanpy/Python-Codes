```python
# 1. Use map() with a lambda to add 5 to every element of the following nested list [[1, 
# 2], [3, 4], [5, 6]] 
l = [[1,2],[3,4],[5,6]]
s = map(lambda x: [i+5 for i in x], l)
print(list(s))
```

```python
# 2. Given a dictionary: d = {"apple": 100, "banana": 40, "cherry": 150} . Use filter() to 
# keep only the keys whose values are greater than 50. 
d = {"apple":100, "banana":40, "cherry":150}

s = list(filter(lambda k: d[k] > 50, d.keys()))
print(s)
```

```python
# 3. Use functools.reduce() with a lambda to find the largest number from a given list 
# Dynamically.  
from functools import reduce

n = list(map(int, input("Enter numbers: ").split()))

s = reduce(lambda a, b: a if a > b else b, n)

print("Maximum:", s)
```

```python
# 4. What happens if the lambda passed to reduce() accepts only one parameter or 
# three parameters? Explain the output or error.  
from functools import reduce
l=[1,2,3,4]
s=reduce(lambda x :x , l)
print(list(s))
```

```python
# 5. Use map() on a string to convert each character into its ASCII value (using ord()). 
# Print the result list.
text="kethan"
s=map(lambda x:ord(x), text)
print(list(s))
```

```python
# 6. Use filter() to remove all vowels from a string and print the final string. 
k = "kethan"
s = filter(lambda i: i not in "aeiou", k)
print(list(s))
```

```python
# 7. Use reduce() to concatenate a list of characters into a single string. Example input: 
['P', 'y', 't', 'h', 'o', 'n'].  
from functools import reduce
p=['P', 'y', 't', 'h', 'o', 'n']
s=reduce(lambda a,b: a+b,p)
print(s)
```

```python
# 10. Given a list of numbers: [5, 10, 15, 20, 25, 30] Perform the following in a single 
# pipeline:  
# • Use map() to square each number  
# • Use filter() to keep only numbers divisible by 5  
# • Use reduce() to calculate the sum of remaining numbers
from functools import reduce

num=[5, 10, 15, 20, 25, 30] 
s=reduce(lambda a,b:a+b ,filter(lambda i : i%5==0, map(lambda x :x**2 , num)))
print(s)
```

```python
# 11. Explain the difference between map(), filter(), and reduce() in Python.  
# • What does each function return?  
# • When should you prefer lambda functions over normal functions?  
# map()
nums = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, nums))

print(result)

# filter()
nums = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, nums))

print(result)
# reduce()
from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda x, y: x + y, nums)

print(result)

```

```python
# 12. Given two lists: a = [1, 2, 3, 4] b = [10, 20, 30, 40] Use map() with a lambda to 
# create a new list containing the sum of corresponding elements. What happens if the 
# lists are of unequal length?  
a = [1, 2, 3, 4]
b = [10, 20, 30, 40] 
s=list(map(lambda x,y : x+y, a,b))
print(s)
```

```python
# 13. Given a list: nums = [12, 15, 7, 18, 20, 21, 25] Use filter() and lambda to keep 
# numbers that are divisible by 3 OR divisible by 5 but NOT divisible by both. Explain 
# how the logical condition works. 
nums = [12, 15, 7, 18, 20, 21, 25]

result = list(filter(lambda x: (x % 3 == 0) ^ (x % 5 == 0), nums))

print(result)
 
```
