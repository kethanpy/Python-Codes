````python
# Q11. Create a class Student that: 
# • Keeps track of the total number of students created. 
# • Determines whether a student passed or failed based on a shared passing mark. 
# • Provides a method to curve marks by increasing everyone’s marks by a percentage. 
# • Has a utility to convert marks (0–100) into letter grades (A, B, C, etc.). 
# Demonstrate: 
# 1. Creating multiple students. 
# 2. Applying a grading curve. 
# 3. Displaying updated results with letter grades. 

class Student:

    # Class variables
    total_students = 0
    passing_mark = 40

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

        Student.total_students += 1

    # Instance method
    def result(self):
        if self.marks >= Student.passing_mark:
            status = "Pass"
        else:
            status = "Fail"

        grade = Student.grade(self.marks)

        print(self.name, self.marks, status, grade)

    # Class method - grading curve
    @classmethod
    def curve_marks(cls, percentage):
        # Increase every student's marks
        for student in cls.students:
            student.marks += student.marks * percentage / 100

            # Maximum marks = 100
            if student.marks > 100:
                student.marks = 100

    # Static method - convert marks to grade
    @staticmethod
    def grade(marks):
        if marks >= 90:
            return "A"
        elif marks >= 75:
            return "B"
        elif marks >= 60:
            return "C"
        elif marks >= 40:
            return "D"
        else:
            return "F"


# Store all students
Student.students = []

# Create students
s1 = Student("Kethan", 35)
s2 = Student("Ravi", 65)
s3 = Student("Suresh", 80)

Student.students.extend([s1, s2, s3])

print("Before Curve:")
s1.result()
s2.result()
s3.result()

# Apply 10% grading curve
Student.curve_marks(10)

print("\nAfter 10% Curve:")
s1.result()
s2.result()
s3.result()

print("\nTotal Students:", Student.total_students)
````

````python
# Q17. Build an Inventory class that: 
# • Tracks the total number of items across all inventories. 
# • Each instance maintains its own stock dictionary ({"item": quantity}). 
# • Provides a method to add or remove stock. 
# • Allows updating a minimum stock threshold globally. 
# • Offers a static checker to verify if a stock level is below threshold. 
# Demonstrate: 
# 1. Managing multiple inventories. 
# 2. Adjusting stock threshold. 
# 3. Using static validation inside the instance logic. 

class Inventory:

    # Class variables
    total_items = 0
    min_threshold = 5

    def __init__(self):
        # Each object has its own stock dictionary
        self.stock = {}

    # Add stock
    def add_stock(self, item, quantity):
        self.stock[item] = self.stock.get(item, 0) + quantity
        Inventory.total_items += quantity

    # Remove stock
    def remove_stock(self, item, quantity):
        if item in self.stock and self.stock[item] >= quantity:
            self.stock[item] -= quantity
            Inventory.total_items -= quantity
        else:
            print("Not enough stock!")

    # Change minimum threshold globally
    @classmethod
    def set_threshold(cls, threshold):
        cls.min_threshold = threshold

    # Static checker
    @staticmethod
    def is_below_threshold(quantity):
        return quantity < Inventory.min_threshold

    # Display inventory
    def show_stock(self):
        print("Stock:", self.stock)

        for item, quantity in self.stock.items():
            if Inventory.is_below_threshold(quantity):
                print(item, "is below minimum stock")


# -------------------------
# Creating multiple inventories
# -------------------------

inventory1 = Inventory()
inventory2 = Inventory()

# Inventory 1
inventory1.add_stock("Laptop", 10)
inventory1.add_stock("Mouse", 3)

# Inventory 2
inventory2.add_stock("Keyboard", 8)
inventory2.add_stock("Mouse", 6)

print("Inventory 1:")
inventory1.show_stock()

print("\nInventory 2:")
inventory2.show_stock()

print("\nTotal Items:", Inventory.total_items)


# -------------------------
# Remove stock
# -------------------------

inventory1.remove_stock("Laptop", 2)

print("\nAfter removing 2 Laptops:")
inventory1.show_stock()

print("Total Items:", Inventory.total_items)


# -------------------------
# Change threshold globally
# -------------------------

Inventory.set_threshold(7)

print("\nNew Threshold:", Inventory.min_threshold)

print("\nInventory 1 after threshold change:")
inventory1.show_stock()

print("\nInventory 2 after threshold change:")
inventory2.show_stock()
````
