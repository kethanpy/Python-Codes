```python
# 1. Write a function that accepts a list and returns the sum of even numbers.

def sum_even(numbers):
    total = 0

    for num in numbers:
        if num % 2 == 0:
            total += num

    return total


numbers = [1, 2, 3, 4, 5, 6]
result = sum_even(numbers)

print(result)
```

```python
# 2. Create a function that takes a string and returns the number of vowels.

def count_vowels(text):
    count = 0

    for ch in text:
        if ch in "aeiouAEIOU":
            count += 1

    return count


text = "Hello World"
result = count_vowels(text)

print(result)
```

```python

# 3. Write a function using **kwargs to print student details.

def student_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)


student_details(
    name="Kethan",
    age=22,
    branch="ECE",
    year=2025
)

```

```python
# 4. Write a function where inner function modifies outer variable using nonlocal

def outer():
    count = 0

    def inner():
        nonlocal count
        count += 1
        print("Count:", count)

    inner()
    inner()
    inner()


outer()
```

```python
# 5. Convert list of temperatures in Celsius to Fahrenheit using map().


def celsius_to_fahrenheit(c):
    return (c * 9 / 5) + 32


celsius = [0, 10, 20, 30, 40]

fahrenheit = list(map(celsius_to_fahrenheit, celsius))

print(fahrenheit)
```

```python
# 7. Use map() with lambda to extract first letters of names.
names = ["Kethan", "Ravi", "Suresh", "Anil"]

first_letters = list(map(lambda name: name[0], names))

print(first_letters)
```

```python

# 8. Filter strings that start with vowel.

words = ["apple", "banana", "orange", "grape", "umbrella", "mango"]

vowel_words = list(filter(lambda word: word[0].lower() in "aeiou", words))

print(vowel_words)
```

```python
# 9. Multiply only even numbers using reduce().
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

result = reduce(lambda x, y: x * y, even_numbers)

print(result)
```

```python
# 10. Compute factorial of a number using reduce().

from functools import reduce

n = 5

result = reduce(lambda x, y: x * y, range(1, n + 1))

print(result)
```

```python
# 11. Use filter() + reduce() to sum only odd numbers.

from functools import reduce

numbers = [1, 2, 3, 4, 5, 6, 7]

odd_numbers = filter(lambda x: x % 2 != 0, numbers)

result = reduce(lambda x, y: x + y, odd_numbers)

print(result)
```
