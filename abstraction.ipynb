```python
"""12. Design an abstract class PaymentGateway with: 
• authenticate() 
• pay(amount) 
• refund(amount) 
Implement subclasses: 
• UPIPayment 
• CardPayment 
• NetBankingPayment 
Show how abstraction helps your main program call payment methods without caring about 
the payment type. """
from abc import ABC, abstractmethod

class PaymentGateway(ABC):

    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class UPIPayment(PaymentGateway):

    def authenticate(self):
        print("Your UPI PIN is correct")

    def pay(self, amount):
        print(f"{amount} is paid through UPI")

    def refund(self, amount):
        print(f"{amount} is refunded through UPI")


class CardPayment(PaymentGateway):

    def authenticate(self):
        print("Your Card PIN is correct")

    def pay(self, amount):
        print(f"{amount} is paid through Card")

    def refund(self, amount):
        print(f"{amount} is refunded through Card")


class NetBankingPayment(PaymentGateway):

    def authenticate(self):
        print("Net Banking Login Successful")

    def pay(self, amount):
        print(f"{amount} is paid through Net Banking")

    def refund(self, amount):
        print(f"{amount} is refunded through Net Banking")


def payment_type(payment, amount):
    payment.authenticate()
    payment.pay(amount)


upi = UPIPayment()
card = CardPayment()
netbank = NetBankingPayment()

payment_type(upi, 200)

payment_type(card, 300)

payment_type(netbank, 500)

upi.refund(50)
card.refund(100)
netbank.refund(200)
    
        
         

        
        


    
    
```

```python
"""19. Create an abstract MediaPlayer with: 
• load() 
• play() 
• stop() 
Implement: 
• MP3Player 
• WAVPlayer 
• AACPlayer 
Demonstrate calling each via a unified interface. """

from abc import ABC,abstractmethod
class mediaplayer(ABC):
    @abstractmethod
    def load(self):
        pass
    @abstractmethod
    def play(self):
        pass
    @abstractmethod
    def stop(self):
        pass
class MP3player(mediaplayer):
    def load(self):
        print("mp3player is loading")
    def play(self):
        print("mp3player is paly")
    def stop(self):
        print("mp3player is stop")
class WAVPlayer(mediaplayer):
    def load(self):
        print("wavplayer is load")
    def play(self):
        print("wavplayer is play")
    def stop(self):
        print("wavplayer is stop")
class AACplayer(mediaplayer):
    def load(self):
        print("accplayer is load")
    def play(self):
        print("accplayer is play")
    def stop(self):
        print("accplayer is stoping")
        
        
def play_un(play):

    play.load()
    play.play()
    play.stop()

    mp3 = MP3player()
    wp = WAVPlayer()
    ap = AACplayer()

mp3 = MP3player()
wp = WAVPlayer()
ap = AACplayer()

play_un(mp3)
play_un(wp)
play_un(ap)
```

```python
"""20. Design: 
• Abstract base class Sensor with functions read_value() and calibrate() 
• Subclasses: TemperatureSensor, PressureSensor, HumiditySensor 
Encapsulate: 
• internal raw sensor readings 
• calibration factor 
Hide all raw operations and allow only a public, clean get_reading() method."""


from abc import ABC, abstractmethod


class Sensor(ABC):

    def __init__(self, raw_value, calibration_factor=1.0):
        self.__raw_value = raw_value
        self.__calibration_factor = calibration_factor

    @abstractmethod
    def read_value(self):
        pass

    @abstractmethod
    def calibrate(self):
        pass

    def get_reading(self):
        value = self.read_value()
        return value


class TemperatureSensor(Sensor):

    def read_value(self):
        return self._Sensor__raw_value * self._Sensor__calibration_factor

    def calibrate(self):
        self._Sensor__calibration_factor = 1.05


class PressureSensor(Sensor):

    def read_value(self):
        return self._Sensor__raw_value * self._Sensor__calibration_factor

    def calibrate(self):
        self._Sensor__calibration_factor = 0.98


class HumiditySensor(Sensor):

    def read_value(self):
        return self._Sensor__raw_value * self._Sensor__calibration_factor

    def calibrate(self):
        self._Sensor__calibration_factor = 1.02


temperature = TemperatureSensor(25)
pressure = PressureSensor(1000)
humidity = HumiditySensor(60)

temperature.calibrate()
pressure.calibrate()
humidity.calibrate()

print(temperature.get_reading())
print(pressure.get_reading())
print(humidity.get_reading())
```
